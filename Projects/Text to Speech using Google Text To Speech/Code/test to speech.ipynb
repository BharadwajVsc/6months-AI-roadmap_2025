```python
import pyttsx3
from IPython.display import Audio

text = gTTS("Admiration... is the furthest thing from understanding. "
    "You look at power and see glory… I look at it — and see truth. "
    "You were never fighting me... only the illusion I allowed you to believe in. "

    "The weak seek justice. The strong… create it. And I? "
    "I simply shape the world... to reflect what it has always been — mine. "

    "You speak of loyalty. Of bonds. But tell me — when everything crumbles... "
    "will those ideals protect you... or shatter... with you? "

    "Victory... was never uncertain. It was merely delayed. For my amusement. "

    "You fear what you cannot comprehend. That... is the limit of your evolution... "
    "and the beginning... of mine. "

    "Do not mistake mercy... for weakness. If I allowed you to live... "
    "it was because your death... would be inconvenient. "

    "Even gods must kneel... eventually. The difference is... I choose when. "

    "It must be exhausting... pretending your actions have meaning. "
    "I’ve already won. I’m only waiting for you... to realize it.")

audio = pyttsx3.init()
audio.setProperty('rate', 150)
audio.setProperty('volume', 0.8)

# change the voices
voice = audio.getProperty('voices')

# 0 for male and 1 for female
audio.setProperty('voice', voice[0].id)      # for male voice   
#audio.setProperty('voice', voice[1].id)       # for female voice

# text-to speech conversion
audio.say(text)

# save the audio file
audio.save_to_file(text, 'test_male_Voice.mp3')
#audio.save_to_file(text, 'test_female_Voice.mp3')

audio.runAndWait()

```

```python
import pyttsx3

text = (
    "Admiration... is the furthest thing from understanding. "
    "You look at power and see glory… I look at it — and see truth. "
    "You were never fighting me... only the illusion I allowed you to believe in. "
    "The weak seek justice. The strong… create it. And I? "
    "I simply shape the world... to reflect what it has always been — mine. "
    "You speak of loyalty. Of bonds. But tell me — when everything crumbles... "
    "will those ideals protect you... or shatter... with you? "
    "Victory... was never uncertain. It was merely delayed. For my amusement. "
    "You fear what you cannot comprehend. That... is the limit of your evolution... "
    "and the beginning... of mine. "
    "Do not mistake mercy... for weakness. If I allowed you to live... "
    "it was because your death... would be inconvenient. "
    "Even gods must kneel... eventually. The difference is... I choose when. "
    "It must be exhausting... pretending your actions have meaning. "
    "I’ve already won. I’m only waiting for you... to realize it."
)

engine = pyttsx3.init()
engine.setProperty('rate', 150)      # Speed of speech
engine.setProperty('volume', 0.9)    # Volume (0.0 to 1.0)

# Choose voice
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[0].id)  # Change index for male/female

# Speak or save
engine.save_to_file(text, 'aizen_offline.mp3')
engine.runAndWait()
print("Saved as 'aizen_offline.mp3'")
```
