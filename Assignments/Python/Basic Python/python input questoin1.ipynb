8. Write a Python program that accepts a string and prints the number of vowels in it. 


```python
s = input('Enter a string: ')
vowels = 'aeiouAEIOU'
vowels_count = 0
for c in s:
    if c in vowels:
        vowels_count+= 1
print(f'The string is: {s}')
print(f'The number of vowels in the string is: {vowels_count}')
```

16. Write a program that checks if the input number is a prime number.

```python
a = int(input("Enter a number: "))

if a <= 1:
    print(f"{a} is not a prime number.")
else:
    for i in range(2, a):
        if a % i == 0:
            print(f"{a} is not a prime number.")
            break
    else:
        print(f"{a} is a prime number.")
```
