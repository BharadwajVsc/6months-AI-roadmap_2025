# Masking

```python
mat
```

```python
id(mat)
```

```python
mat<50 # will give a boolean matrix where the values are True if the value is less than 50 and False otherwise
```

```python
mat>50 # will give a boolean matrix where the values are True if the value is greater than 50 and False otherwise
```

```python
mat == 50 # will give a boolean matrix where the values are True if the value is equal to 50 and False otherwise
```

```python
mat[mat == 50] # will give the values that are equal to 50
```

```python
x = mat[mat < 50] # will give the values that are less than 50
x
```

```python
y = mat[mat > 50] # will give the values that are greater than 50
y
```

```python
a = mat[mat <= 50] # will give the values that are less than or equal to 50
a
```
