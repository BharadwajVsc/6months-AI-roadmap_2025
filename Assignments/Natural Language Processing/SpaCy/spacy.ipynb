```python
import spacy 

nlp = spacy.load('en_core_web_sm')
text = 'Apple is looking at buying U.K. startup for $1 billion'
doc = nlp(text)# while using spacy, we need to create a doc object to store the processed text
```

```python
doc
```

```python
#printing names entities found in the text
print('Named Entities, Phrases and Concepts:')
for ent in doc.ents:
    print(f'{ent.text:15} {ent.label_:10} {ent.start_char:10}')
```
