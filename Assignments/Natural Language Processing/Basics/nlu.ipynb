```python
from nltk.stem import PorterStemmer
pst=PorterStemmer()
```

```python
pst.stem('running')
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
words_to_stem=['running', 'affection', 'playing', 'maximum', 'happily', 'happiness', 'unhappily', 'unhappiness', 'asking', 'paying', 'gives']
for words in words_to_stem:
    print(words, ":", pst.stem(words))
```

```python
words_to_stem=['give', 'given', 'giving', 'gives', 'gave', 'paying', 'paid', 'pay', 'pays', 'bharadwaj']
for words in words_to_stem:
    print(words, ":", pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst=LancasterStemmer() # LancasterStemmer is a more aggressive stemmer than PorterStemmer, it often produces shorter stems. Example: 'given' becomes 'giv', 'paying' becomes 'pay'.

words_to_stem=['give', 'given', 'giving', 'gives', 'gave', 'paying', 'paid', 'pay', 'pays', 'bharadwaj']
for words in words_to_stem:
    print(words, ":", lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer
ss=SnowballStemmer('english') # SnowballStemmer is a more advanced stemmer that supports multiple languages and provides better stemming results than PorterStemmer.

for words in words_to_stem:
    print(words, ":", ss.stem(words))
```

```python
ss=SnowballStemmer('german')# Example of using SnowballStemmer for German language
words_to_stem_german=['laufen', 'laufen', 'laufend', 'lief', 'gelaufen', 'läuft', 'laufst', 'lauft']
for words in words_to_stem_german:
    print(words, ":", ss.stem(words))
```

```python
from nltk import ne_chunk
```
