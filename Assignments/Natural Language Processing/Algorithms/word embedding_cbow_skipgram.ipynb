# **NLP Algorithms: CBOW, Skipgram and Word2Vec**

### Word2Vec:
- will find the semantic meaning of the tokens and group the token with similar meaning. 
- will convert the words in to numerical values based on their meaning.
- these words are converted into vectors and are stored in `vector db`.
- is divided into:
    - CBOW
    - Skipgram
- `Gensim` : wil fetch the semantic meaning of the words with the vectors.
- Advantages:
    - Efficiency: it can efficiently process large datasets and learn high quality word embeddings.
    - Capturing Semantic Relationships: captures semantic relationship between words better than One-Hot Coding.
- Limitations:
    - Static Representations: generates static embeddings, where each word will have a single vector representation regardless of its context. This can cause issue with polysemy words.
    - `Out-Of-Vocabulary Words(OOV)`: words that are not int he training dataset will not have embeddings which will limit the model's ability to generalize. 

```python
import gensim
from gensim.models import Word2Vec

sentences = [
    ['the', 'cat', 'sat', 'on', 'the', 'mat'],
    ['the', 'dog', 'sat', 'on', 'the', 'log'],
    ['the', 'cat', 'and', 'the', 'dog', 'are', 'friends'],
    ['the', 'mat', 'is', 'soft'],
    ['the', 'log', 'is', 'hard']
]

# train words2vec model
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg = 1)

#getting vector for a word
vector = model.wv['cat']
print("Vector for 'cat':", vector)

#finding similar words
similar_words = model.wv.most_similar('cat', topn=3)
print("Most similar words to 'cat':", similar_words)
```

### CBow and Skpigram
- Continuous Bag Of Words(CBOW): will predict the target word based on the context of surrounding words.
    - works well with small datasets
    

- Skipgram: it predicts the surrounding context word using the target word.
    - is more effective on large datasets. 
    - captures more info about the relationships between the words which makes it useful for understanding the semantics.

- for cbow model, `sg=0` is used and `sg=1` is used for skipgram model
- `vector_size` is the length of each word vector
    - larger the value, it capture more semantic details but requires more computation and data
    - smaller values are faster , require less memory but may miss fine grained relationships
    - Common values used: 100, 200, 300
- `window` is the distance between the current word and the next contextual word
    - smaller window will capture syntactical/grammatical relations
    - larger window will capture sematic/contextual relations


```python
import gensim
from gensim.models import Word2Vec

sentences = [
    ['I', 'love', 'natural', 'language', 'processing'],
    ['I', 'enjoy', 'learning', 'about', 'machine', 'learning'],
    ['Natural', 'language', 'processing', 'is', 'fascinating'],
    ['Machine', 'learning', 'is', 'a', 'subset', 'of', 'artificial', 'intelligence'],
    ['I', 'like', 'to', 'explore', 'deep', 'learning']
]

#cbow model
cb_model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, sg=0) #sg=0 is for CBOW

#skipgram model
skipgram_model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, sg=1)

#getting word for a word
cb_vector = cb_model.wv['language']
skipgram_vector = skipgram_model.wv['language']
print("CBOW Vector for 'language':", cb_vector)
print("Skipgram Vector for 'language':", skipgram_vector)

#finding similar words
cb_similar_words = cb_model.wv.most_similar('language', topn=3)
skipgram_similar_words = skipgram_model.wv.most_similar('language', topn=5)
print("CBOW Most similar words to 'language':", cb_similar_words)
print("Skipgram Most similar words to 'language':", skipgram_similar_words)
```
