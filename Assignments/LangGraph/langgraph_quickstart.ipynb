```python
from langgraph.graph import StateGraph, END
from typing import TypedDict
from IPython.display import Image, display
```

```python
class MyState(TypedDict):
    message: str

#defining the starting node
def start_node(state: MyState)-> MyState: #-> is used to show the return type. Here, the function takes MyState as input and returns MyState as output.
    print('Node has started:', state['message'])
    #process or pass the state to the end node
    return{'message': state['message'] + '-> processed by start_node'}

#defining the end node
def end_node(state: MyState)-> MyState:
    print('Node has ended:', state['message'])
    return state

#creating the state graph
builder=StateGraph(MyState)
```

```python
builder=StateGraph(MyState)
```

```python
#Adding nodes to the graph
builder.add_node('start', start_node)
builder.add_node('end', end_node)
```

```python
#Defining the flow

builder.set_entry_point('start')
builder.add_edge('start', 'end')
builder.add_edge('end', END)#END is a special node that indicates the end of the graph
```

```python
#Compiling the graph
graph=builder.compile()

#run the graph
result= graph.invoke({'message': 'Hello, LangGraph!'})
print('Final State:', result['message'])
```
