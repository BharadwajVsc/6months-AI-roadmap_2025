# **Data Augmentation** 

## Import necessary libraries

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array

```

## Setting Up the hyperparameters as per requirement

```python
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    #fill_mode='nearest',
    fill_mode='wrap',
    #fill_mode='reflect',
    #fill_mode='constant',
    rescale=None
)

```

## Loading the image

```python
img=load_img(r'D:\adobe lightroom\DSC06098.jpg')
img
```

## Convert Image to Array

```python
x = img_to_array(img)
x =x.reshape((1,)+x.shape) #reshaping to (1, height, width, channels)
```

## Generating Augmented Image and Saving It

```python
i=0
for batch in datagen.flow(x, batch_size=1,
                        save_to_dir=r"D:\fsds\practice_codes\deep_learning\augmented images",
                        save_prefix='bike',
                        save_format='jpeg'):
    i=i+1
    if i>10:
        break
```
