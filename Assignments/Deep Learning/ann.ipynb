# **Artificial Neural Network(ANN)**

## Import necessary libraries 

```python
import numpy as np 
import pandas as pd
import tensorflow as tf
```

```python
tf.__version__
```

## PHASE 1: Data Preprocessing

```python
dataset = pd.read_csv(r'D:\fsds\practice_codes\deep_learning\Churn_Modelling.csv')
X = dataset.iloc[:, 3:-1].values
y = dataset.iloc[:, -1].values

dataset.head()
```

### Applying Label Encoder on `Gender` column



```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
X[:,2]=le.fit_transform(X[:,2])

print(X)
```

##### Applying One Hot Encoding on `Geography` column 

```python
from sklearn.compose import ColumnTransformer #columntransformer is used to apply different transformations to different columns of the data set
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')# remainder = 'passthrough' means all other columns should remain as it is
X = np.array(ct.fit_transform(X))

print(X)
```

### Applying feature scaling 

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)

print(X)
```

### Splitting the dataset into tran and test data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)    
print(y_test.shape)
```

## PHASE 2: Building ANN 

### Initializing ANN

```python
ann = tf.keras.models.Sequential()
```

### Adding input layer


```python
ann.add(tf.keras.layers.Dense(units=6))
```

### Adding input layer and the first hidden layer

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

### Adding second hidden layer

```python
ann.add(tf.keras.layers.Dense(units=5, activation='relu'))
```

### Adding third hidden layer

```python
ann.add(tf.keras.layers.Dense(units=4, activation='relu'))
```

### Adding the output layer

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))#units=1 because we have to predict only one output and activation function is sigmoid because it is a binary classification problem
```

## PHASE 3: Training ANN

### Compiling ANN

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

### Training ANN on training data

```python
ann.fit(X_train, y_train, batch_size = 32, epochs = 100)
```

## PHASE 4: Predicting and evaluating the model

### Test data prediction

```python
y_pred=ann.predict(X_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

### Printing the confusion matrix

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
print(f'Here is the confusion matrix:\n{cm}')
```

### Printing the accuracy score

```python
from sklearn.metrics import accuracy_score
ac=accuracy_score(y_test, y_pred)
print(f'Accuracy of the model is: {ac*100}%')
```
